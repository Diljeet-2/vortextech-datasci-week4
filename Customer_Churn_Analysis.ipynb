# End-to-End Customer Churn Analysis Report

## 1. Executive Summary

In today's highly competitive telecommunications market, retaining existing customers is just as critical as acquiring new ones. This report analyzes historical customer data from a fictional telecommunications company to understand why customers are leaving (churning) and what can be done to keep them. 

Our analysis explored variables such as customer demographics, account tenure, subscribed services, contract types, and billing methods. The findings reveal a clear pattern: customers who are on month-to-month contracts, those with higher monthly charges, and relatively new customers (with short tenure) face the highest risk of leaving. Furthermore, the method of billing is a strong indicator of churn, as customers paying via electronic checks churn at significantly higher rates compared to those using automatic payment methods.

This analysis provides a strong foundation for taking proactive steps. By focusing on transitioning customers to long-term contracts through attractive incentives and monitoring new customers during their crucial first few months, the business can significantly improve retention. Ultimately, reducing churn will lead to increased long-term revenue and higher customer lifetime value, moving the company from a reactive position to a structured, data-driven retention strategy.


## 2. Business Problem

**Primary Question:** Why are customers leaving the company (churning), and what proactive measures can the business take to reduce this customer churn?

By answering this question, the business can design targeted retention campaigns instead of relying on generic marketing, helping it allocate resources effectively to save "at-risk" customers.


## 3. Dataset Overview

The dataset used for this analysis is the **Telco Customer Churn dataset**, representing 7,043 customer profiles. 
Key components include:
*   **Demographics:** Gender, age range (Senior Citizen), partners, and dependents.
*   **Services:** Internet, phone lines, online security, device protection, streaming services, etc.
*   **Account details:** Contract type, payment method, paperless billing, tenure length.
*   **Financials:** Monthly charges and total charges.
*   **Target Variable:** `Churn` (Whether the customer left within the last month).

## 5. Key Findings

Below are the most critical discoveries from our analysis. Each finding directly points to specific behaviors or account features tied to higher churn rates.


### Finding 1: Month-to-Month Contracts Have the Highest Churn Risk
Customers who have not committed to a long-term contract (i.e., those on a month-to-month basis) are vastly more likely to leave the company. This suggests that without the friction of a contract, at-risk customers easily transition to competitors.

![Churn by Contract Type](images/chart2.png)


### Finding 2: Higher Monthly Charges Drive Customers Away
The histogram of monthly charges shows that a significant portion of churning customers are concentrated in the higher billing brackets (specifically above $70/month). Price sensitivity seems to be a major factor in customer defection.

![Monthly Charges vs Churn](images/chart3.png)


### Finding 3: New Customers Are the Most Vulnerable
The majority of customers who left the company did so within their first few months of service. As tenure (months with the company) increases, the likelihood of a customer leaving drops dramatically. 

![Tenure vs Churn](images/chart4.png)


### Finding 4: Electronic Check Users Churn Significantly More
When looking at payment methods, customers who pay manually via "Electronic check" represent a disproportionately large share of the churned user base. In contrast, automated methods (Bank transfer, Credit card) correspond with higher retention.

![Payment Method vs Churn](images/chart5.png)


### Additional Context: Overall Churn Distribution
Currently, approximately 26.6% of the customer base has churned. This baseline tells us how large the problem is across the entire company.

![Churn Distribution](images/chart1.png)


## 6. Recommendations

Based on these findings, the business should implement the following actionable strategies to reduce churn:

1.  **Incentivize Long-Term Contracts:** 
    Since month-to-month customers are the most likely to churn, offer targeted discounts or complimentary service upgrades (e.g., free tech support or expanded streaming) to customers who agree to switch to a 1-year or 2-year contract.
2.  **Launch a "New Customer Care" Campaign:** 
    Given that new customers are the most vulnerable, implement a proactive check-in program during the first 1-3 months. Ensuring they are getting value and resolving early friction points can build long-term loyalty.
3.  **Promote Auto-Pay Options:** 
    Customers using Electronic Checks are leaving at alarming rates. Encourage a transition to automatic payments (Credit Card or Bank Transfer) by offering a small monthly discount (e.g., $5 off) for enabling Auto-Pay. This reduces the manual friction of paying bills and increases retention subconsciously.


## 7. Appendix (Data Cleaning & EDA Code)

The following cells contain all the Python code used to read, clean, and visualize the dataset. Technical stakeholders can review these cells to verify the analytical rigor of our findings.


```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# 1. Load the data
df = pd.read_csv('customer_churn.csv')
df.head()
```

```
# 2. Data Cleaning
# The TotalCharges column is stored as string, we need to convert it to numeric
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

# Check for missing values created by the conversion
print(f"Missing values before dropping: {df['TotalCharges'].isnull().sum()}")

# Drop missing values
df.dropna(inplace=True)
print(f"Shape after cleaning: {df.shape}")
```

```
# 3. Code for Generating Plot 1: Churn Distribution
plt.figure(figsize=(8, 6))
ax = sns.countplot(x='Churn', data=df)
plt.title('Customer Churn Distribution', fontsize=16)
plt.show()
```

```
# 4. Code for Generating Plot 2: Contract Type vs Churn
plt.figure(figsize=(8, 6))
sns.countplot(x='Contract', hue='Churn', data=df)
plt.title('Churn by Contract Type', fontsize=16)
plt.show()
```

```
# 5. Code for Generating Plot 3: Monthly Charges Distribution
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='MonthlyCharges', hue='Churn', kde=True, bins=30)
plt.title('Monthly Charges Distribution by Churn', fontsize=16)
plt.show()
```

```
# 6. Code for Generating Plot 4: Customer Tenure
plt.figure(figsize=(8, 6))
sns.boxplot(x='Churn', y='tenure', data=df)
plt.title('Customer Tenure by Churn Status', fontsize=16)
plt.show()
```

```
# 7. Code for Generating Plot 5: Payment Method
plt.figure(figsize=(10, 6))
sns.countplot(y='PaymentMethod', hue='Churn', data=df)
plt.title('Churn by Payment Method', fontsize=16)
plt.show()
```

```
# 8. Statistical Summary
# Checking correlation between numeric variables and churn
df_corr = df[['tenure', 'MonthlyCharges', 'TotalCharges']].copy()
df_corr['Churn_Binary'] = df['Churn'].map({'Yes': 1, 'No': 0})
sns.heatmap(df_corr.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```
